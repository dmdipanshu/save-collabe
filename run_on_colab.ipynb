### Run Telegram Bot on Google Colab
This single notebook contains all the source code required to run the bot.

```python
import base64
import zipfile
import io
import os

bot_dir = "/content/bot"
os.makedirs(bot_dir, exist_ok=True)
%cd /content/bot

print("Extracting bot code...")
zip_data = ""
with zipfile.ZipFile(io.BytesIO(base64.b64decode(zip_data))) as z:
    z.extractall()
print("Code extracted successfully!")
```
